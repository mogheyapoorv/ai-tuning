```python
from presidio_structured import StructuredEngine, JsonAnalysisBuilder, PandasAnalysisBuilder, StructuredAnalysis, CsvReader, JsonReader, JsonDataProcessor, PandasDataProcessor
```

```python
sample_csv_data = CsvReader().read('./sample_data/test_structured.csv')
sample_csv_data
```

```python
sample_json_data = JsonReader().read('./sample_data/test_structured.json')
sample_json_data
```

```python
sample_complex_json_data = JsonReader().read('./sample_data/test_structured_complex.json')
sample_complex_json_data
```

```python
tabular_analysis = PandasAnalysisBuilder().generate_analysis(sample_csv_data)
tabular_analysis
```

```python
pandas_engine = StructuredEngine(data_processor=PandasDataProcessor())
df_to_anonymize = sample_csv_data.copy()
anonymize_df = pandas_engine.anonymize(df_to_anonymize, tabular_analysis, operators=None)
anonymize_df
```
