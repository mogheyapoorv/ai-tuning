```python
from presidio_analyzer import AnalyzerEngine
from presidio_analyzer.nlp_engine import SpacyNlpEngine

text = "Hi, My name is Aoorva Moghey, my email id moghey.apoorv@gmail.com and my phone number is 9910060220."

nlp_engine = SpacyNlpEngine(models=[{'lang_code': 'en', 'model_name': 'en_core_web_lg'}])

analyser = AnalyzerEngine(nlp_engine=nlp_engine)
analyser_result = analyser.analyze(text=text, language="en")

print(f"Analysis with SpaCy recognizer (formatted output)")
for result in analyser_result:
    print(f" Entity Type: {result.entity_type}, Start: {result.start}, End: {result.end}, Score: {result.score:.2f}")
```

```python
from presidio_analyzer import AnalyzerEngine
from presidio_analyzer.nlp_engine import SpacyNlpEngine
from presidio_anonymizer import AnonymizerEngine

# Define the text to be analyzed and anonymized
text = "Hi, My name is Aoorva Moghey, my email id abcx.xyzf@gmail.com and my phone number is 9999999999."

# Download the model
!python -m spacy download en_core_web_lg
# Create a SpaCyNlpEngine with the downloaded model
nlp_engine = SpacyNlpEngine(models=[{'lang_code': 'en', 'model_name': 'en_core_web_lg'}])

# Initialize the AnalyzerEngine with the custom NLP engine
analyser_with_spacy = AnalyzerEngine(nlp_engine=nlp_engine)

# Re-run analysis with the new engine
analyser_result_spacy = analyser_with_spacy.analyze(text=text, language="en")

print("Analysis with SpaCy recognizer (formatted output):")
for result in analyser_result_spacy:
    print(f"  Entity Type: {result.entity_type}, Start: {result.start}, End: {result.end}, Score: {result.score:.2f}")

anonymizer = AnonymizerEngine()
anonymizer_text = anonymizer.anonymize(text=text, analyzer_results=analyser_result_spacy)

print("\nAnonymized Text:")
print(anonymizer_text.text)
```

```python
from presidio_anonymizer import AnonymizerEngine

anonymizer = AnonymizerEngine()
anonymized_text = anonymizer.anonymize(text=text, analyzer_results=analyser_result_spacy)

print("Original Text:")
print(text)
print("\nAnonymized Text:")
print(anonymized_text.text)
```
